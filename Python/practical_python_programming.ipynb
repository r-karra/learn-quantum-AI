```python
principal = 500000.0
rate = 0.05
payment = 2684.11
extra_payment_amount = 1000.0
extra_payment_months = 12
total_paid = 0.0
month_count = 0

print(f"Starting principal: {principal}")
print(f"Monthly payment: {payment}")
print(f"Extra payment for first {extra_payment_months} months: {extra_payment_amount}")

while principal > 0:
    month_count += 1
    current_payment = payment
    if month_count <= extra_payment_months:
        current_payment += extra_payment_amount

    # Calculate interest for the month
    interest_payment = principal * (rate / 12)
    #print(interest_payment)
    principal = principal + interest_payment - current_payment

    total_paid += current_payment

    # Prevent principal from going negative beyond the last payment
    if principal <= 0:
        total_paid += principal # Subtract the overpayment
        principal = 0

    print(f"Month {month_count}: Remaining principal = {principal:.2f}, Paid this month = {current_payment:.2f}")

print('\nTotal paid:', round(total_paid, 2))
print('Total months:', month_count)
```
